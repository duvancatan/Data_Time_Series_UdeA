<h1 align="center"><b> Periodograma - Índice de Oscilación del Sur (SOI) </b></h1>

---
##### El Índice de Oscilación del Sur (IOS), o SOI, es un indicador climatológico clave que mide las fluctuaciones de la presión atmosférica superficial entre el Pacífico Occidental (Darwin, Australia) y el Pacífico Central/Oriental (Tahití). Se calcula estandarizando la diferencia de presión entre estas dos estaciones. Los valores negativos persistentes del IOS reflejan un debilitamiento de la presión en Tahití y vientos alisios débiles, lo que está fuertemente asociado con las condiciones de El Niño (aguas cálidas en el Pacífico Oriental); mientras que los valores positivos persistentes indican una intensificación de los vientos alisios y una mayor diferencia de presión, característicos del fenómeno de La Niña (aguas frías). El IOS, por lo tanto, sirve para cuantificar la fase e intensidad de la interacción océano-atmósfera conocida como ENSO (El Niño-Oscilación del Sur).

📈 **Serie de tiempo SOI en la librería astsa**

- **Nombre:** SOI

- **Tipo de objeto:** ts (serie de tiempo en R)

- **Frecuencia:** Mensual

- **Periodo:** De enero de 1950 a diciembre de 2000 (es decir, 612 observaciones mensuales)

- **Unidad:** Índice estandarizado (sin unidades físicas directas)

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

🔍 **¿Qué indica el SOI?**

- Un valor negativo del SOI suele estar asociado con eventos El Niño, caracterizados por condiciones más cálidas en el océano Pacífico oriental.

- Un valor positivo del SOI suele indicar un evento La Niña, con condiciones más frías en el Pacífico oriental.

- Valores cercanos a cero implican condiciones neutrales.

```r
# Crear variable de tiempo #
t <- 1:length(soi)

# Ajustar regresión lineal #
modelo_trend <- lm(soi ~ t)

# Resumen del modelo #
summary(modelo_trend)
```

```r
# ==============================================================================
# 1. PREPARACIÓN DE DATOS Y AJUSTE DEL MODELO
# ==============================================================================
# Crear variable de tiempo secuencial t
t <- 1:length(soi)

# Ajustar regresión lineal sobre t
modelo_trend <- lm(soi ~ t)

# Construir data.frame con el índice t, los años reales y los valores ajustados
df_soi <- data.frame(
  t         = t,
  Tiempo    = as.numeric(time(soi)),
  SOI       = as.numeric(soi),
  Tendencia = predict(modelo_trend)
)

# ==============================================================================
# 2. GRÁFICO PROFESIONAL EN GGPLOT2
# ==============================================================================
g_soi <- ggplot(df_soi, aes(x = Tiempo)) +
  # Línea de referencia horizontal en y = 0
  geom_hline(yintercept = 0, color = "gray50", linetype = "dashed", linewidth = 0.5) +
  
  # Serie temporal principal (Azul marino institucional)
  geom_line(aes(y = SOI, color = "Serie SOI"), linewidth = 0.55, alpha = 0.85) +
  
  # Línea de tendencia calculada sobre t (Naranja bermellón)
  geom_line(aes(y = Tendencia, color = "Línea de Tendencia"), linewidth = 1.1) +
  
  # Asignación manual de colores de alto contraste
  scale_color_manual(
    values = c("Serie SOI" = "#1F4E79", "Línea de Tendencia" = "#E64A19")
  ) +
  
  # Escala del eje X basada en el tiempo real (marcas cada 5 años)
  scale_x_continuous(breaks = seq(floor(min(df_soi$Tiempo)), ceiling(max(df_soi$Tiempo)), by = 5)) +
  
  # Títulos y etiquetas
  labs(
    title = "Índice de Oscilación del Sur (SOI) y Tendencia Lineal",
    subtitle = paste0("Modelo de regresión: SOI ~ t | Pendiente (β1) = ", 
                       formatC(coef(modelo_trend)[2], format = "e", digits = 3)),
    x = "Año",
    y = "Índice SOI",
    color = "Componente:"
  ) +
  
  # Estética minimalista y limpia
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(face = "bold", size = 16, hjust = 0.5),
    plot.subtitle = element_text(size = 12, hjust = 0.5, color = "gray30"),
    axis.title = element_text(face = "bold", size = 12),
    legend.position = "bottom",
    legend.title = element_text(face = "bold", size = 11),
    panel.grid.minor = element_blank(),
    panel.grid.major = element_line(color = "gray88", linewidth = 0.4)
  )

print(g_soi)
```

```r
# ==============================================================================
# 1. EXTRACCIÓN DE RESIDUOS (DESTENDENCIALIZACIÓN) Y DATOS
# ==============================================================================
# Extraer residuos del modelo de tendencia lineal
soi_dest <- resid(modelo_trend)

# Construir data.frame con el tiempo real y los residuos
df_soi_dest <- data.frame(
  Tiempo   = as.numeric(time(soi)),
  Residuo  = as.numeric(soi_dest)
)

# ==============================================================================
# 2. GRÁFICO PROFESIONAL EN GGPLOT2
# ==============================================================================
g_soi_dest <- ggplot(df_soi_dest, aes(x = Tiempo, y = Residuo)) +
  # Línea de referencia horizontal en y = 0
  geom_hline(yintercept = 0, color = "#E64A19", linetype = "dashed", linewidth = 0.7) +
  
  # Serie temporal destendencializada (Azul marino)
  geom_line(color = "#1F4E79", linewidth = 0.55, alpha = 0.85) +
  
  # Escala del eje X con saltos cada 5 años
  scale_x_continuous(breaks = seq(floor(min(df_soi_dest$Tiempo)), ceiling(max(df_soi_dest$Tiempo)), by = 5)) +
  
  # Títulos y etiquetas formalizadas
  labs(
    title = "Serie SOI Destendencializada (Residuos de Regresión Lineal)",
    subtitle = "Componente oscilatorio sin tendencia lineal | Centrada en cero (y = 0)",
    x = "Año",
    y = "SOI (Detrended)"
  ) +
  
  # Estética minimalista
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(face = "bold", size = 16, hjust = 0.5),
    plot.subtitle = element_text(size = 12, hjust = 0.5, color = "gray30"),
    axis.title = element_text(face = "bold", size = 12),
    panel.grid.minor = element_blank(),
    panel.grid.major = element_line(color = "gray88", linewidth = 0.4)
  )

print(g_soi_dest)
```

```r
# Cálculo del Periodograma #
I <- abs(fft(soi_dest))^2 / 444    # Periodograma
P <- (4 / 444) * I[1:222]          # Periodograma escalado
f <- 0:221 / 444                   # Frecuencias
```

```r
# Gráfico del periodograma #
plot(f, P, type = "l",
     xlab = "Frecuencia",
     ylab = "Periodograma Escalado",
     main = "Periodograma de la serie SOI")
```

2. Calcular el periodograma de la serie sin tendencia obtenida anteriormente. Identifica las frecuencias de los cuatro picos principales (con uno evidente en la frecuencia de un ciclo cada 12 meses) ¿Cuál es el probable ciclo de El Niño indicado por el pico menor?

</div>

---

```r
# 1. Encontrar el índice del valor máximo en P $
indice_max_P <- which.max(P)

# 2. Usar ese índice para encontrar el valor de frecuencia f correspondiente #
frecuencia_maxima <- f[indice_max_P]

# 3. Mostrar el resultado #
print(paste("El valor máximo de P se encuentra en el índice:", indice_max_P))
print(paste("La frecuencia f correspondiente (ciclos/unidad) es:", frecuencia_maxima))

# 4. Calcular el Periodo # 
periodo_maximo <- 1 / frecuencia_maxima
print(paste("El Periodo (unidades de tiempo) es:", periodo_maximo))
```

```r
# 1. Ordenar los valores de P de mayor a menor y obtener los índices de los 5 más grandes
# 'decreasing = TRUE' ordena de mayor a menor.
# 'head(..., 5)' toma los primeros 5 elementos.
indices_top_5 <- head(order(P, decreasing = TRUE), 5)

# 2. Obtener las frecuencias y potencias correspondientes
frecuencias_top_5 <- f[indices_top_5]
potencias_top_5 <- P[indices_top_5]

# 3. Crear una tabla de resultados
resultados_maximos <- data.frame(
  Indice = indices_top_5,
  Frecuencia_f = frecuencias_top_5,
  Potencia_P = potencias_top_5,
  Periodo = 1 / frecuencias_top_5
)

print(resultados_maximos)
```

```r
# 1. Preparación de los datos: Identificar y calcular el Top 4
num_maximos <- 4

# Ordenar los valores de P y obtener los índices de los 3 más grandes
indices_top_4 <- head(order(P, decreasing = TRUE), num_maximos)

# Obtener los valores de f, P y calcular el Periodo para esos 3 picos
frecuencias_top_4 <- f[indices_top_4]
potencias_top_4 <- P[indices_top_4]
periodos_top_4 <- 1 / frecuencias_top_4

# 2. Generar la Gráfica del Periodograma (P vs f)
plot(f, P,
     type = "l",
     main = "Periodograma con los 4 Picos de Potencia más Altos",
     xlab = "Frecuencia Normalizada (f, Ciclos/Unidad)",
     ylab = "Densidad Espectral de Potencia (P)",
     col = "darkblue",
     lwd = 2)

# 3. Resaltar los 4 máximos (Puntos rojos)
points(frecuencias_top_4, potencias_top_4,
       col = "red",
       pch = 19,
       cex = 1.3) # Un poco más grande para destacarlos

# 4. Añadir las Etiquetas de Texto (Tripletas: (f, P, Periodo))
for (i in 1:length(frecuencias_top_4)) {

  # Formateamos el texto de la tripleta (f, P, Periodo)
  etiqueta <- paste0("(",
                     round(frecuencias_top_4[i], 3), ", ",  # Frecuencia (3 decimales)
                     round(potencias_top_4[i], 2), ", ",    # Potencia (2 decimales)
                     round(periodos_top_4[i], 1),           # Periodo (1 decimal)
                     ")")

  # Agregamos el texto al gráfico
  # pos = 4 coloca el texto a la derecha del punto
  text(frecuencias_top_4[i], potencias_top_4[i],
       labels = etiqueta,
       pos = 4,
       cex = 0.85,         # Tamaño de fuente
       col = "darkred")
}

# 5. Añadir la Leyenda
legend("topright",
       legend = c("Periodograma", "Top 4 Picos", "(f, P, Periodo)"),
       col = c("darkblue", "red", "darkred"),
       lwd = c(2, NA, NA),
       pch = c(NA, 19, NA),
       bty = "n")
```
